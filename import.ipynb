```python
import pandas as pd
```

```python
df=pd.read_csv('exp 4.csv')
```

```python
df
```

```python
df1=df.pivot_table(index='type',columns='department',values='salary')
print(df1)
```

```python
df1=df.pivot_table(index='type',columns='department',values='salary',aggfunc='mean')
print(df1)
```

```python
df2=df.pivot_table(index='department',columns='type',values='salary',)
print(df2)
```

```python
df1=df.pivot_table(index='type',columns='department',values='salary',aggfunc=['mean','sum','count'])
print(df1)
```

```python
df1=df.pivot_table(index='type',columns='department',values='salary',aggfunc='std')
print(df1)
```
